```python
import pandas as pd

df = pd.read_csv("tfg_corpus_final.csv", dtype={"id": str, "identifier": str})

# Marca todas las filas cuyo texto aparece mas de una vez (keep=False marca todas las copias)
dup_mask = df["text"].duplicated(keep=False)
duplicates = df[dup_mask].sort_values("text")

print(f"Filas totales: {len(df)}")
print(f"Filas con text repetido: {len(duplicates)}")
print(f"Textos distintos que estan repetidos: {duplicates['text'].nunique()}")

duplicates
```

```python
# Resumen: cada texto repetido con la cantidad de veces que aparece
conteo = (
    duplicates["text"]
    .value_counts()
    .rename_axis("text")
    .reset_index(name="veces")
)
conteo
```

```python
total = conteo['veces'].sum()
total
```

```python
# Deduplicado: elimina textos repetidos y conserva la primera incidencia
df_dedup = df.drop_duplicates(subset="text", keep="first")

print(f"Filas antes: {len(df)}")
print(f"Filas despues: {len(df_dedup)}")
print(f"Filas eliminadas: {len(df) - len(df_dedup)}")
print(f"Textos duplicados restantes: {df_dedup['text'].duplicated().sum()}")

df_dedup.to_csv("tfg_corpus_final_dedup.csv", index=False)
```
